```python
import pygame
import numpy as np
from realtime_heart_plot import RealtimeHeartPlot
from workload_speedometer import WorkloadSpeedometer

pygame.init()
screen = pygame.display.set_mode((1000, 700))
pygame.display.set_caption("HR + WL Modular Display")

# === Proper initialization ===
heart_plot = RealtimeHeartPlot(position=(50, 50), hr_interval_seconds=5)
workload_meter = WorkloadSpeedometer(850, 500)

# === Main loop (STRICT: 1 line each) ===
running = True
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    screen.fill((255, 255, 255))

    heart_plot.update(); heart_plot.render(screen)                                     # ONE LINE for HR
    workload_meter.update(np.random.rand()); workload_meter.render(screen, (850, 500)) # ONE LINE for WL

    pygame.display.flip()

pygame.quit()
```
